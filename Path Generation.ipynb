```python
try:
  import pandas as pd
  import numpy as np
  import re
  import pickle
  from xgboost import XGBClassifier
  from sklearn.model_selection import train_test_split
  from sklearn.preprocessing import LabelEncoder
except Exception as e:
  print(e)
```

```python
# Load Dataset
file_path = "/content/drive/MyDrive/AI/job_data_1.csv"  # Update path if needed
df = pd.read_csv(file_path)
```

```python
# Remove unwanted columns
df = df.drop(columns=["Uniq Id", "Crawl Timestamp","Job Salary","Location","Functional Area","Industry"], errors='ignore')
```

```python
# Function to extract minimum experience from text
def extract_experience(exp):
    if isinstance(exp, str):
        match = re.search(r'(\d+)', exp)
        return int(match.group(1)) if match else np.nan
    return np.nan
```

```python
df["Job Experience Required"] = df["Job Experience Required"].apply(extract_experience)
```

```python
df["Experience Level"] = pd.cut(
    df["Job Experience Required"],
    bins=[-1, 2, 5, float("inf")],
    labels=["Entry", "Mid", "Senior"]
)
```

```python
df["Job Title"] = df["Job Title"].fillna("Unknown")
df["Key Skills"] = df["Key Skills"].fillna("")
```

```python
# Selecting features and target
features = df[["Job Title", "Key Skills"]]
target = df["Experience Level"].dropna()
```

```python
# Convert categorical features to numerical
features_encoded = pd.get_dummies(features["Job Title"], drop_first=True, prefix='job_title') # Add prefix
features_encoded = features_encoded.loc[target.index]
```

```python
features_encoded.columns = features_encoded.columns.str.replace('[\[\]\<]', '', regex=True)
```

```python
# Create a LabelEncoder object
label_encoder = LabelEncoder()
```

```python
# Function to recommend required skills for a job title
def recommend_skills(job_title):
    relevant_jobs = df[df["Job Title"].str.contains(job_title, case=False, na=False)]
    skills = relevant_jobs["Key Skills"].str.split(", ").explode().value_counts().index.tolist()
    return {f"Skill {i+1}": skill for i, skill in enumerate(skills)}
```

```python
# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(features_encoded, target, test_size=0.2, random_state=42)
```

```python
y_train = label_encoder.fit_transform(y_train)
y_test = label_encoder.transform(y_test) # Also transform y_test
```

```python
# Train an XGBoost Classifier
model = XGBClassifier(eval_metric='mlogloss', random_state=42)
model.fit(X_train, y_train)
```

```python
# Save the trained model as a .pkl file
with open("/content/drive/MyDrive/AI/career_model.pkl", "wb") as file:
    pickle.dump(model, file)

print("Model saved successfully as career_model.pkl")
```

```python
def recommend_skills(job_title):
    relevant_jobs = df[df["Job Title"].str.contains(job_title, case=False, na=False)]
    skills = relevant_jobs["Key Skills"].str.split(", ").explode().value_counts().index.tolist()
    return {f"Skill {i+1}": skill for i, skill in enumerate(skills)}
```

```python
# Function to predict experience level
def predict_experience_level(job_title):
    # Create a DataFrame for the new job title
    new_data = pd.DataFrame({"Job Title": [job_title]})

    # One-hot encode the job title
    # Ensure the columns match those used during training
    # Get all the job titles used during training
    all_job_titles = features_encoded.columns.str.replace('job_title_', '').tolist()
    new_data_encoded = pd.get_dummies(new_data["Job Title"], prefix='job_title')
    # Add missing columns if any
    for title in all_job_titles:
        if f'job_title_{title}' not in new_data_encoded.columns:
            new_data_encoded[f'job_title_{title}'] = 0
    # Ensure the order of columns is the same as in training data
    new_data_encoded = new_data_encoded[[f'job_title_{title}' for title in all_job_titles]]

    # Predict using the loaded model
    prediction = loaded_model.predict(new_data_encoded)[0]

    # Decode the prediction (if LabelEncoder was used)
    experience_level = label_encoder.inverse_transform([prediction])[0]

    return experience_level
```
